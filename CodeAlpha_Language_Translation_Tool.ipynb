```python
# Create a safe translation function
def translate_text(text, source_language, target_language):

    # Check if the input is empty
    if not text.strip():
        return "Please enter some text."

    try:
        # Create the translator
        translator = GoogleTranslator(
            source=source_language,
            target=target_language
        )

        # Translate the text
        translated = translator.translate(text)

        # Return the translation
        return translated

    except Exception as error:

        # Return a simple error message
        return "Translation failed. Please try another sentence."
```

```python
# Create a translation function with error handling
def translate_text(text, source_language, target_language):

    # Check if the text is empty
    if not text.strip():
        return "Please enter some text to translate."

    try:
        # Translate the text
        translated = GoogleTranslator(
            source=source_language,
            target=target_language
        ).translate(text)

        # Return the translated text
        return translated

    except Exception:
        # Show a message if translation fails
        return "Translation failed. Please try again."
```

```python
# Store the English text
text = "Good morning"

# Store language names and their language codes
languages = {
    "Telugu": "te",
    "Hindi": "hi",
    "Tamil": "ta",
    "Kannada": "kn"
}

# Translate the text into each language
for language, code in languages.items():

    # Translate the text
    result = translate_text(text, "en", code)

    # Display the translation
    print(language + ":", result)
```

```python
# Store supported languages and their codes
languages = {
    "English": "en",
    "Telugu": "te",
    "Hindi": "hi",
    "Tamil": "ta",
    "Kannada": "kn",
    "Malayalam": "ml",
    "Bengali": "bn",
    "Marathi": "mr"
}

# Display the available languages
print(languages)
```

```python
# Get the names of all supported languages
language_names = list(languages.keys())

# Display the language names
print(language_names)
```

```python
# Create a function to get the language code
def get_language_code(language_name):

    # Return the code of the selected language
    return languages[language_name]
```

```python
# Get the language code for Telugu
code = get_language_code("Telugu")

# Display the language code
print("Telugu code:", code)
```

```python
# Get the language code for Hindi
code = get_language_code("Hindi")

# Display the language code
print("Hindi code:", code)
```

```python
# Create a function to translate using language names
def translate_by_language(text, source_language, target_language):

    # Convert source language name into its code
    source_code = get_language_code(source_language)

    # Convert target language name into its code
    target_code = get_language_code(target_language)

    # Translate the text
    result = translate_text(text, source_code, target_code)

    # Return the translated text
    return result
```

```python
# Store the text
text = "Good morning"

# Translate English into Telugu
result = translate_by_language(
    text,
    "English",
    "Telugu"
)

# Display the result
print("Original:", text)
print("Translated:", result)
```

```python
# Import Gradio
import gradio as gr

# Import Google Translator
from deep_translator import GoogleTranslator


# Store language names and their codes
languages = {
    "English": "en",
    "Telugu": "te",
    "Hindi": "hi",
    "Tamil": "ta",
    "Kannada": "kn",
    "Malayalam": "ml",
    "Bengali": "bn",
    "Marathi": "mr"
}


# Create the translation function
def translate_by_language(text, source_language, target_language):

    # Check if the input box is empty
    if not text.strip():
        return "Please enter some text."

    try:
        # Get the source language code
        source_code = languages[source_language]

        # Get the target language code
        target_code = languages[target_language]

        # Translate the text
        translated_text = GoogleTranslator(
            source=source_code,
            target=target_code
        ).translate(text)

        # Return the translated text
        return translated_text

    except Exception as error:

        # Return an error message
        return f"Translation error: {error}"


# Create the Gradio application
with gr.Blocks() as app:

    # Add the title
    gr.Markdown("# 🌐 Language Translation Tool")

    # Add the description
    gr.Markdown(
        "Translate text from one language to another."
    )

    # Create the source language dropdown
    source_language = gr.Dropdown(
        choices=list(languages.keys()),
        label="Source Language",
        value="English"
    )

    # Create the input text box
    input_text = gr.Textbox(
        label="Enter Text",
        placeholder="Type your text here..."
    )

    # Create the target language dropdown
    target_language = gr.Dropdown(
        choices=list(languages.keys()),
        label="Target Language",
        value="Telugu"
    )

    # Create the Translate button
    translate_button = gr.Button("Translate")

    # Create the output box
    # The copy icon is built into this box
    output_text = gr.Textbox(
        label="Translated Text",
        buttons=["copy"]
    )

    # Create the Clear button
    clear_button = gr.Button("Clear")

    # Connect the Translate button
    translate_button.click(
        fn=translate_by_language,
        inputs=[
            input_text,
            source_language,
            target_language
        ],
        outputs=output_text
    )

    # Connect the Clear button
    clear_button.click(
        fn=lambda: ("", "English", "Telugu", ""),
        inputs=[],
        outputs=[
            input_text,
            source_language,
            target_language,
            output_text
        ]
    )


# Display a message
print("Language Translation Tool is ready!")
```

```python
# Launch the application in the browser
app.launch(inbrowser=True)
```
